# Logistic Regression — The Library Version

Same model, ~5 lines of scikit-learn, verifying our scratch build.

*(One expected difference: sklearn regularizes by default (`C=1.0`) — README §4/§6 explain why that's a feature. We set `C` very high to make the comparison with our unregularized scratch build fair.)*

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix, precision_score, recall_score, accuracy_score

df = pd.read_csv("data/exam_data.csv")
X = df[["hours_studied", "prev_score"]]
y = df["passed"]

Xs = StandardScaler().fit_transform(X)          # Block 3 of the scratch notebook
model = LogisticRegression(C=1e6)               # C huge = (almost) no regularization, like our scratch build
model.fit(Xs, y)                                # replaces Blocks 4–7

print(f"sklearn learned:  w = {model.coef_[0].round(3)}  b = {model.intercept_[0]:.3f}")
print( "our scratch run:  w = [2.946 1.965]  b = 0.414  ← same answer; ours stopped a few steps\n"
       "                    earlier on the flat bottom of the bowl (more epochs → identical)")
```

```python
pred = model.predict(Xs)
print("Confusion matrix:\n", confusion_matrix(y, pred))
print(f"Accuracy  = {accuracy_score(y, pred):.2f}")
print(f"Precision = {precision_score(y, pred):.2f}")
print(f"Recall    = {recall_score(y, pred):.2f}")
# Same numbers as Block 9 of the scratch notebook — README §5 explains how to read each.
```

**The takeaway:** `LogisticRegression.fit()` is our Blocks 4–7 in professional packaging: a sigmoid on a line, cross-entropy going downhill, plus default regularization for the perfectly-separable trap (README §6). No magic — you built it.
