# Naive Bayes — The Library Version

Same model via scikit-learn's `CountVectorizer` (our Block 3) + `MultinomialNB` (our Blocks 4–6), verifying the scratch build.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score

df = pd.read_csv("data/sms_data.csv")
tr, te = train_test_split(df, test_size=0.2, random_state=21)

vec = CountVectorizer()
Xtr = vec.fit_transform(tr.message)      # fit vocabulary on TRAIN only — README §9
Xte = vec.transform(te.message)          # test words outside it are dropped, like our Block 3

model = MultinomialNB(alpha=1.0)         # alpha = our Laplace smoothing knob (README §4)
model.fit(Xtr, tr.label)                 # 'fit' = counting. Instant.

pred = model.predict(Xte)
print("Confusion matrix (rows truth: ham, spam):")
print(confusion_matrix(te.label, pred, labels=["ham", "spam"]))
print(f"Accuracy  = {accuracy_score(te.label, pred):.0%}")
print(f"Precision = {precision_score(te.label, pred, pos_label='spam'):.0%}")
print(f"Recall    = {recall_score(te.label, pred, pos_label='spam'):.0%}")
```

**The takeaway:** `MultinomialNB.fit()` is two count tables and a prior — our Blocks 4–5 — and `alpha` is the same +1 smoothing we hand-rolled. The instant fit is the tell: counting, not optimizing (README §3.3).
