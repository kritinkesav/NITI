# Linear Regression — The Library Version

In practice you'll use scikit-learn. This notebook shows the model in ~5 lines **and verifies it agrees with our from-scratch build** — proof that the library is doing exactly what you implemented, just faster and battle-tested.

*(sklearn's `LinearRegression` uses the closed-form/least-squares solution — the "jump straight to the bottom of the bowl" shortcut from README §3.4's footnote — so tiny differences vs gradient descent are expected and should be negligible.)*

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

df = pd.read_csv("data/salary_data.csv")
X = df[["years_experience"]]        # sklearn wants a 2-D table of features
y = df["salary"]

model = LinearRegression()
model.fit(X, y)                     # this replaces our entire training loop

print(f"sklearn learned:  salary ≈ {model.coef_[0]:,.0f} × years + {model.intercept_:,.0f}")
print( "our scratch run:  salary ≈ ~5,2xx × years + ~30,xxx   ← should match closely")
```

```python
pred = model.predict(X)
print(f"RMSE = {np.sqrt(mean_squared_error(y, pred)):,.0f}")
print(f"MAE  = {mean_absolute_error(y, pred):,.0f}")
print(f"R²   = {r2_score(y, pred):.3f}")
# Same metrics as Block 9 of the scratch notebook — README §5 explains how to read each.
```

**The takeaway:** the library isn't magic — it's Blocks 4–7 of the scratch notebook, professionally packaged. Now when `model.fit()` misbehaves on a real project, you know what's inside: a loss being minimized, gradients (or algebra) finding the bottom of a bowl, and knobs you understand.
