# Neural Network — The Library Version

Same architecture via scikit-learn's `MLPClassifier`, verifying the scratch build — plus the map to the tools the deep-learning world actually uses (PyTorch/TensorFlow) and what they add.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score

df = pd.read_csv("data/moons_data.csv")
X = df[["x1", "x2"]].values
y = (df.label == "B").astype(int).values
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.2, random_state=88)
sc = StandardScaler().fit(Xtr)                      # train-only stats — the sealed exam
Xtr_s, Xte_s = sc.transform(Xtr), sc.transform(Xte)

mlp = MLPClassifier(hidden_layer_sizes=(8,), activation="tanh",   # our exact architecture
                    solver="sgd", learning_rate_init=0.5, max_iter=4000,
                    random_state=0)
mlp.fit(Xtr_s, ytr)
print(f"MLP (8 tanh hidden) test accuracy: {accuracy_score(yte, mlp.predict(Xte_s)):.0%}  <- matches the scratch build")
print(f"parameters learned: {sum(w.size for w in mlp.coefs_) + sum(b.size for b in mlp.intercepts_)}  (our 33)")
```
